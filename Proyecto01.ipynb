##Generando archivos PDF del presupuesto
*Crear archivo PDF
*Definir el tipo de letra
*Llenar los datos
*Guardar el entregable

```python
%pip install fpdf
```

```python
from fpdf import FPDF
from google.colab import files
import ipywidgets as widgets
from IPython.display import display, clear_output
import os

# Widgets del formulario
campo_proyecto = widgets.Text(description="Proyecto:", placeholder="Descripción del proyecto")
campo_horas = widgets.Text(description="Horas:", placeholder="Ej: 40")
campo_valor_hora = widgets.Text(description="Valor/hora:", placeholder="Ej: 20000")
campo_termino = widgets.Text(description="Término:", placeholder="Ej: 2 semanas")
boton_generar = widgets.Button(description="Generar Presupuesto", button_style="success")
salida = widgets.Output()

def generar_pdf(b):
    with salida:
        clear_output()

        proyecto = campo_proyecto.value.strip()
        horas_texto = campo_horas.value.strip()
        valor_hora_texto = campo_valor_hora.value.strip()
        termino = campo_termino.value.strip()

        # Validaciones
        if proyecto == "" or proyecto.isdigit():
            print("La descripción del proyecto no puede estar vacía ni ser solo números.")
            return
        if not horas_texto.isdigit() or int(horas_texto) <= 0:
            print("Las horas estimadas deben ser un número entero positivo.")
            return
        if not valor_hora_texto.isdigit() or int(valor_hora_texto) <= 0:
            print("El valor de la hora debe ser un número entero positivo.")
            return
        if termino == "":
            print("El término estimado no puede estar vacío.")
            return

        horas_estimadas = int(horas_texto)
        valor_hora = int(valor_hora_texto)
        valor_total = horas_estimadas * valor_hora

        if not os.path.exists("Template.png"):
            print("Falta el archivo Template.png. Súbelo con files.upload() y vuelve a generar.")
            return

        pdf = FPDF()
        pdf.add_page()
        pdf.set_font("Arial")
        pdf.image("Template.png", x=0, y=0)

        pdf.text(115, 145, proyecto)
        pdf.text(115, 160, str(horas_estimadas))
        pdf.text(115, 175, str(valor_hora))
        pdf.text(115, 190, termino)
        pdf.text(115, 205, str(valor_total))

        pdf.output("Presupuesto.pdf")
        print("¡Presupuesto generado exitosamente!")
        files.download("Presupuesto.pdf")

boton_generar.on_click(generar_pdf)

# Subir el template antes de mostrar el formulario, si no existe
if not os.path.exists("Template.png"):
    print("Sube el archivo Template.png:")
    uploaded = files.upload()

display(campo_proyecto, campo_horas, campo_valor_hora, campo_termino, boton_generar, salida)
```
